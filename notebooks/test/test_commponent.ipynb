### 1️⃣ تحميل البيئة والمفاتيح (Environment Setup)
نستخدم مكتبة  لتحميل المفاتيح تلقائياً من ملف  الموجود في جذر المشروع.

</div>

```python
import os
from dotenv import load_dotenv, find_dotenv

# تحميل البيئة تلقائياً من ملف .env
load_dotenv(find_dotenv())

# التحقق من وجود مفتاح Groq
api_key = os.getenv("GROQ_API_KEY")
if api_key:
    print("✅ تم العثور على GROQ_API_KEY بنجاح!")
else:
    print("⚠️ لم يتم العثور على GROQ_API_KEY. يرجى التأكد من وجوده في ملف .env")
```

<div dir="rtl">

### 2️⃣ تهيئة نموذج ChatGroq
نقوم باستدعاء  مباشرة وتحديد نموذج Groq مثل  دون أي تعقيد.

</div>

```python
from langchain_groq import ChatGroq

# تهيئة نموذج Groq بطريقة بسيطة ومباشرة
llm = ChatGroq(
    model = "openai/gpt-oss-120b",
    temperature=0.7
)

print("🚀 تم تهيئة النموذج بنجاح!")
print(f"اسم النموذج: {llm.model}")
```

<div dir="rtl">

### 3️⃣ إرسال سؤال واستلام الإجابة (Invocation)
نستخدم الدالة  وتأخذ السؤال مباشرة وتُرجع كائن إجابة نستخرج منه النص عبر .

</div>

```python
# إرسال سؤال مباشر للنموذج
question = "who is the best club in the world history say one club"
response = llm.invoke(question)

print("questtion", question)
print("model answer")
print(response.content)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful AI assistant."),
    ("human", "{question}")
])
```

```python
chain = prompt|llm
responce = chain.invoke("input can you tell me about langsmit")
print(responce)
```

```python
from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()

chain = prompt | llm | output_parser

response = chain.invoke({
    "question": "Can you tell me about LangSmith?"
})

print(response)
```
