# ⚡ 02 - FastEmbed (ONNX) in LangChain

## ما هو `FastEmbed`؟
- مكتبة تضمينات فائقة السرعة وموفرة للذاكرة من Qdrant.
- تعتمد على محرك **ONNX Runtime** ولا تحتاج لتشغيل أطر عمل ثقيلة مثل PyTorch بالكامل.
- تدعم نماذج قوية ومحدثة مثل `BAAI/bge-small-en-v1.5`.

</div>


```python
from langchain_community.embeddings.fastembed import FastEmbedEmbeddings

# تهيئة نموذج FastEmbed
fast_embeddings = FastEmbedEmbeddings(
    model_name="BAAI/bge-small-en-v1.5"
)

text = "FastEmbed is optimized for high-throughput embedding generation."
vector = fast_embeddings.embed_query(text)

print(f"✅ أبعاد متجه FastEmbed: {len(vector)}")
print(f"عينة من المتجه: {vector[:5]}")
```
