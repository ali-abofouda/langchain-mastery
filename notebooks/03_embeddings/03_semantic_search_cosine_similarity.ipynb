# 🔍 03 - Semantic Search & Cosine Similarity

## البحث الدلالي وحساب التشابه
- **Cosine Similarity**: قياس جيب تمام الزاوية بين متجهين في الفضاء المتجهي.
- القيمة تتراوح بين `0` (لا يوجد تشابه) و `1.0` (تطابق تام في المعنى).
- يتيح العثور على النصوص الأكثر صلة بالسؤال حتى وإن لم تستخدم نفس الكلمات الحرفية.

</div>


```python
import numpy as np
from langchain_huggingface import HuggingFaceEmbeddings

hf_embeddings = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2",
    encode_kwargs={"normalize_embeddings": True}
)

def cosine_similarity(vec_a, vec_b):
    a = np.array(vec_a)
    b = np.array(vec_b)
    return float(np.dot(a, b) / (np.linalg.norm(a) * np.linalg.norm(b)))

# مجموعة مستندات تجريبية
corpus = [
    "Deep learning uses artificial neural networks with multiple layers.",
    "A delicious recipe for authentic Italian pizza and pasta.",
    "Convolutional Neural Networks are primarily used for image classification.",
    "The match ended with a thrilling 3-2 victory."
]

corpus_embeddings = hf_embeddings.embed_documents(corpus)

# سؤال المستخدم
user_query = "How are neural networks used to analyze images and computer vision?"
query_embedding = hf_embeddings.embed_query(user_query)

# حساب درجة التطابق لكل مستند
ranked_results = []
for i, doc_emb in enumerate(corpus_embeddings):
    score = cosine_similarity(query_embedding, doc_emb)
    ranked_results.append((corpus[i], score))

ranked_results.sort(key=lambda x: x[1], reverse=True)

print(f"🔍 سؤال المستخدم: '{user_query}'\n")
print("📊 النتائج مرتبة حسب التطابق الدلالي:")
print("-" * 65)
for rank, (doc, score) in enumerate(ranked_results, 1):
    print(f"[{rank}] Score: {score:.4f} | {doc}")
```
