# 🧠 01 - HuggingFace Local Embeddings in LangChain

## ما هي التضمينات (Embeddings)؟
- تحويل النصوص إلى متجهات عددية (Numerical Vectors) في فضاء دلالي لفهم معاني وسياق النصوص.
- نموذج `all-MiniLM-L6-v2`:
  - يعمل محلياً 100% بدون أي مفاتيح API وبدون تكلفة.
  - يُنتج متجهاً من **384 بعداً (Dimensions)**.
- **الدوال الأساسية**:
  - `embed_query`: لتضمين استعلام بحث مفرد.
  - `embed_documents`: لتضمين قائمة نصوص دفعة واحدة.

</div>


<div dir="rtl">

### ⚙️ تحميل مفاتيح البيئة من ملف `.env` (`HF_TOKEN`)
نقوم باستخدام `find_dotenv()` للبحث عن ملف الـ `.env` في المجلد الرئيسي وتحميل المفاتيح تلقائياً لتوثيق الاتصال مع Hugging Face.

</div>


```python
import os
from dotenv import load_dotenv, find_dotenv

# البحث التلقائي عن ملف .env وتحميله
load_dotenv(find_dotenv())

# التحقق من وجود التوكين
hf_token = os.getenv("HF_TOKEN") or os.getenv("HUGGINGFACEHUB_API_TOKEN")
if hf_token and not hf_token.startswith("your_"):
    print("✅ تم تحميل Hugging Face Token وتفعيله بنجاح!")
else:
    print("ℹ️ يعمل النموذج الآن بدون توكين (النماذج العامة تعمل مجاناً وبدون مشاكل).")
```

<div dir="rtl">

### 1️⃣ تهيئة وتجربة النموذج (`HuggingFaceEmbeddings`)

</div>


```python
from langchain_huggingface import HuggingFaceEmbeddings

# 1. تهيئة نموذج HuggingFace محلياً
hf_embeddings = HuggingFaceEmbeddings(
    model_name="all-MiniLM-L6-v2",
    model_kwargs={"device": "cpu"},
    encode_kwargs={"normalize_embeddings": True}
)

# 2. تضمين استعلام مفرد
query = "What is artificial intelligence?"
query_vec = hf_embeddings.embed_query(query)
print(f"✅ أبعاد متجه الاستعلام: {len(query_vec)}")
print(f"عينة من أول 5 أرقام في المتجه: {query_vec[:5]}\n")

# 3. تضمين قائمة مستندات
docs = [
    "Machine learning learns from data.",
    "Python is the standard language for AI."
]
doc_vecs = hf_embeddings.embed_documents(docs)
print(f"✅ عدد المتجهات الناتجة: {len(doc_vecs)}")
print(f"أبعاد كل متجه: {[len(v) for v in doc_vecs]}")
```
