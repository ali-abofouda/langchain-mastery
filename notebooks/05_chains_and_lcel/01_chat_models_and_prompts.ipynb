# ⚡ 01 - Chat Models & Chat Prompt Templates

## ما هي نماذج المحادثة (Chat Models) وقوالب التوجيه؟
- **نماذج المحادثة (Chat Models)**: تمثل واجهة التعامل مع النماذج اللغوية الحديثة التي تستقبل قائمة من الرسائل الهيكلية (`SystemMessage`, `HumanMessage`, `AIMessage`) بدلاً من مجرد نصوص خام.
- **منصة Groq**: تقدم استضافة سحابية فائقة السرعة لمعالجة النماذج اللغوية باستخدام رقائق LPUs المتخصصة، مما يوفر سرعات توليد استثنائية وبدون تكلفة للنماذج مفتوحة المصدر.
- **قوالب التوجيه (`ChatPromptTemplate`)**: تتيح بناء وتخصيص رسائل النظام والتعليمات الديناميكية وتمرير المتغيرات بسهولة وأمان.

</div>

<div dir="rtl">

### ⚙️ تهيئة البيئة والمفاتيح (Environment Setup)
تحميل المتغيرات البيئية من ملف `.env` والتحقق من توفر مفتاح `GROQ_API_KEY`.

</div>

```python
import os
from pathlib import Path
from dotenv import load_dotenv, find_dotenv

# تحميل البيئة تلقائياً
load_dotenv(find_dotenv())

# التحقق من وجود مفتاح Groq
groq_api_key = os.getenv("GROQ_API_KEY")
if groq_api_key:
    print("✅ تم العثور على GROQ_API_KEY بنجاح!")
else:
    print("⚠️ تحذير: لم يتم العثور على GROQ_API_KEY في ملف .env")
```

<div dir="rtl">

### 1️⃣ تهيئة نموذج ChatGroq
نقوم باستدعاء فئة `ChatGroq` من مكتبة `langchain_groq` مع تحديد النموذج ومعامل درجة الحرارة `temperature`.

</div>

```python
from langchain_groq import ChatGroq

# تهيئة نموذج Groq بنموذج متقدم وسريع
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature=0.7,
    max_retries=2
)

print("🚀 تم تهيئة النموذج بنجاح!")
print(f"اسم النموذج المعتمد: {llm.model_name}")
```

<div dir="rtl">

### 2️⃣ الاستدعاء المباشر والتعامل مع الرسائل الهيكلية
يمكن استدعاء النموذج مباشرة بسؤال نصي، أو بتمرير قائمة من الرسائل الهيكلية لتوجيه سلوك النموذج.

</div>

```python
from langchain_core.messages import SystemMessage, HumanMessage

# بناء قائمة رسائل هيكلية لتحديد هوية المساعد وسؤال المستخدم
messages = [
    SystemMessage(content="أنت خبير تقني في هندسة الذكاء الاصطناعي وتقدم إجابات موجزة واحترافية باللغة العربية."),
    HumanMessage(content="ما هو الفرق الجوهري بين LLM التقليدي و Chat Model؟")
]

ai_message = llm.invoke(messages)

print(f"نوع المخرج: {type(ai_message)}")
print("\n--- نص الإجابة ---")
print(ai_message.content)
```

<div dir="rtl">

### 3️⃣ بناء قوالب التوجيه الديناميكية (ChatPromptTemplate)
نستخدم `ChatPromptTemplate.from_messages` لبناء قوالب قابلة لإعادة الاستخدام مع متغيرات ديناميكية مثل `{topic}` و `{language}`.

</div>

```python
from langchain_core.prompts import ChatPromptTemplate

# تعريف قالب توجيه ديناميكي
prompt_template = ChatPromptTemplate.from_messages([
    ("system", "You are an elite tutor explaining technical concepts simply in {language}."),
    ("human", "Explain the concept of {topic} in 3 bullet points.")
])

# تنسيق القالب بالمدخلات المطلوبة
formatted_messages = prompt_template.format_messages(
    language="Arabic",
    topic="Vector Embeddings in LangChain"
)

# إرسال الرسائل المنسقة للنموذج
response = llm.invoke(formatted_messages)
print(response.content)
```

<div dir="rtl">

### 4️⃣ البث الحي للمخرجات (Streaming Tokens)
تدعم نماذج `ChatGroq` البث الفوري للرموز النصية أثناء توليدها عبر التابع `.stream()`، مما يحسن تجربة المستخدم.

</div>

```python
prompt = "اذكر لي أهم 3 ميزات لمكتبة LangChain باختصار."
print("📡 استلام الرد عبر البث المباشر (Streaming):\n")

for chunk in llm.stream(prompt):
    print(chunk.content, end="", flush=True)
print("\n\n✅ اكتمل البث بنجاح!")
```
