# 🌐 03 - HTML Text Splitters in LangChain

## تقنيات تقسيم كود وصفحات الـ HTML:
1. **`HTMLHeaderTextSplitter`**:
   - يقسم الصفحة بناءً على وسوم العناوين الهيكلية (`<h1>`, `<h2>`, `<h3>`).
   - يضيف تسلسل العناوين داخل **Metadata** كل قطعة لمعرفة السياق الدقيق للفقرة.
2. **`RecursiveCharacterTextSplitter.from_language(Language.HTML)`**:
   - تقسيم كود الـ HTML مع مراعاة بنية وسوم الـ HTML البرمجية.

</div>


```python
from langchain_text_splitters import HTMLHeaderTextSplitter, RecursiveCharacterTextSplitter, Language

html_string = """
<!DOCTYPE html>
<html>
<body>
    <h1>LangChain Architecture</h1>
    <p>LangChain is a framework for developing applications powered by LLMs.</p>
    <h2>Data Ingestion</h2>
    <p>Load data from PDFs, Webpages, and Databases using Document Loaders.</p>
    <h2>Text Splitting</h2>
    <p>Break large documents into semantically coherent chunks.</p>
</body>
</html>
"""

# 1. التقسيم حسب وسوم العناوين
headers_to_split_on = [
    ("h1", "Header 1"),
    ("h2", "Header 2"),
]
html_header_splitter = HTMLHeaderTextSplitter(headers_to_split_on=headers_to_split_on)
header_splits = html_header_splitter.split_text(html_string)

print(f"✅ عدد القطع من HTML Header Splitter: {len(header_splits)}\n")
for i, doc in enumerate(header_splits, 1):
    print(f"--- Chunk #{i} ---")
    print(f"Content: {doc.page_content}")
    print(f"Metadata: {doc.metadata}\n")

print("=" * 60)

# 2. التقسيم التكراري المخصص للغة HTML
html_rec_splitter = RecursiveCharacterTextSplitter.from_language(
    language=Language.HTML,
    chunk_size=120,
    chunk_overlap=15
)
code_chunks = html_rec_splitter.split_text(html_string)
print(f"✅ عدد القطع من Recursive HTML Splitter: {len(code_chunks)}")
```
