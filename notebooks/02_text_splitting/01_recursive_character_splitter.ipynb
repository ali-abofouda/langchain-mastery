# ✂️ 01 - RecursiveCharacterTextSplitter in LangChain

## لماذا `RecursiveCharacterTextSplitter` هو الخيار الأفضل؟
- يعتمد على التدرج الذكي في الفواصل: `["\n\n", "\n", " ", ""]`.
- يبدأ بمحاولة فصل الفقرات الكبيرة أولاً، فإذا كانت القطعة أكبر من `chunk_size` ينتقل للأسطر، ثم للجمل، ثم للكلمات.
- يضمن الحفاظ على تماسك المعنى السياقي قدر الإمكان.
- يدعم خاصية `chunk_overlap` لمنع انقطاع الفكرة بين القطع المتتالية.

</div>


```python
from pathlib import Path
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

DATA_DIR = Path("data") if Path("data").exists() else Path("../../data") if Path("../../data").exists() else Path("../data")

# 1. تحميل مستند
loader = PyPDFLoader(str(DATA_DIR / "sample.pdf"))
docs = loader.load()

# 2. إعداد الـ Splitter
recursive_splitter = RecursiveCharacterTextSplitter(
    chunk_size=100,
    chunk_overlap=20,
    length_function=len
)

# 3. تقسيم المستندات مع الحفاظ على الـ Metadata
split_docs = recursive_splitter.split_documents(docs)

print(f"✅ عدد القطع الناتجة: {len(split_docs)}\n")
for i, chunk in enumerate(split_docs, 1):
    print(f"--- Chunk #{i} (Length: {len(chunk.page_content)}) ---")
    print(chunk.page_content)
    print(f"Metadata: {chunk.metadata}\n")
```
