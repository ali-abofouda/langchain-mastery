# 🗄️ 04 - RecursiveJsonSplitter in LangChain

## ما هو `RecursiveJsonSplitter`؟
- مخصص للتعامل مع بيانات الـ JSON المتداخلة والمعقدة (مثل استجابات الـ REST APIs وقواعد بيانات NoSQL).
- يقوم بتقسيم ملف الـ JSON لقطع صغيرة دون تجاوز `max_chunk_size`، مع **الحفاظ على المفاتيح الأبوية (Parent Keys)** وهيكل الـ JSON سليماً في كل قطعة.

</div>


```python
import json
from langchain_text_splitters import RecursiveJsonSplitter

json_data = {
    "company": "AI Solutions Corp",
    "department": {
        "name": "Research and Development",
        "projects": [
            {
                "project_id": "PRJ-001",
                "title": "Enterprise RAG System",
                "status": "Active",
                "budget": 50000
            },
            {
                "project_id": "PRJ-002",
                "title": "Autonomous Agents",
                "status": "In Progress",
                "budget": 75000
            }
        ]
    }
}

json_splitter = RecursiveJsonSplitter(max_chunk_size=180)

# 1. التقسيم كقوائم من JSON Dictionaries
json_chunks = json_splitter.split_json(json_data=json_data)
print(f"✅ عدد قطع الـ JSON: {len(json_chunks)}\n")
for i, chunk in enumerate(json_chunks, 1):
    print(f"--- JSON Chunk #{i} ---")
    print(json.dumps(chunk, indent=2))

print("\n" + "=" * 60)

# 2. التحويل المباشر إلى LangChain Documents
json_docs = json_splitter.create_documents(texts=[json_data])
print(f"✅ عدد مستندات الـ Documents الناتجة: {len(json_docs)}")
print("محتوى أول مستند Document:")
print(json_docs[0].page_content)
```
