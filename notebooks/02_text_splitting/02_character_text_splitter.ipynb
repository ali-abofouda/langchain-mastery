# 📏 02 - CharacterTextSplitter in LangChain

## ما هو `CharacterTextSplitter`؟
- أبسط أنواع المقسمات: يعتمد على **فاصل محدد وثابت** فقط (مثل `\n\n` أو مسافة).
- لا يتدرج بين فواصل متعددة مثل الـ Recursive Splitter، بل يقسم فقط عند الفاصل الممرر.

</div>


```python
from langchain_text_splitters import CharacterTextSplitter

raw_text = """LangChain is an AI framework.

It provides tools for building RAG applications.

It makes chunking and embeddings straightforward."""

char_splitter = CharacterTextSplitter(
    separator="\n\n",
    chunk_size=50,
    chunk_overlap=0
)

chunks = char_splitter.split_text(raw_text)

print(f"✅ عدد القطع الناتجة: {len(chunks)}\n")
for i, c in enumerate(chunks, 1):
    print(f"Chunk #{i}: {c}")
```
