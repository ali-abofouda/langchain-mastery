# 🌐 03 - WebBaseLoader (Web Scraping) in LangChain

## ما هو `WebBaseLoader`؟
- أداة قوية لجلب واستخلاص المحتوى النصي من صفحات الويب والمواقع الإلكترونية.
- تعتمد داخلياً على `urllib` ومكتبة `BeautifulSoup4`.
- تتيح فلترة عناصر الـ HTML غير المرغوبة (مثل الإعلانات والـ Navbars) عبر `SoupStrainer` لاستهداف النص المفيد فقط.

</div>


```python
import bs4
from langchain_community.document_loaders import WebBaseLoader

# استهداف المحتوى الرئيسي لمقال ويكيبيديا فقط
target_url = "https://en.wikipedia.org/wiki/Artificial_intelligence"
loader = WebBaseLoader(
    web_paths=(target_url,),
    bs_kwargs=dict(
        parse_only=bs4.SoupStrainer(id="mw-content-text")
    )
)

web_docs = loader.load()

print(f"✅ المصدر: {web_docs[0].metadata['source']}")
print(f"📊 إجمالي عدد الأحرف: {len(web_docs[0].page_content)}")
print("\n--- معاينة من أول 400 حرف ---")
print(web_docs[0].page_content[:400].strip())
```
