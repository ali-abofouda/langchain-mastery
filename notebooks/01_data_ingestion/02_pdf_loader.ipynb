# 📑 02 - PyPDFLoader in LangChain

## ما هو `PyPDFLoader`؟
- قارئ مخصص لملفات الـ PDF يعتمد على مكتبة `pypdf`.
- **ميزته الأساسية**: يقوم بتقسيم ملف الـ PDF تلقائياً بحيث تكون **كل صفحة في كائن `Document` مستقل**.
- الـ `metadata` لكل صفحة تحتوي على:
  - `page`: رقم الصفحة (0-indexed).
  - `source`: مسار الملف الأصلي.

</div>


```python
from pathlib import Path
from langchain_community.document_loaders import PyPDFLoader

DATA_DIR = Path("data") if Path("data").exists() else Path("../../data") if Path("../../data").exists() else Path("../data")
pdf_path = DATA_DIR / "sample.pdf"

# تحميل ملف الـ PDF
loader = PyPDFLoader(str(pdf_path))
pages = loader.load()

print(f"✅ إجمالي عدد الصفحات: {len(pages)}")
print("--- محتوى الصفحة الأولى ---")
print(pages[0].page_content)
print("\n--- بيانات الصفحة الوصفية (Metadata) ---")
print(pages[0].metadata)
```
