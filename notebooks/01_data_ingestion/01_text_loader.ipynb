# 📄 01 - TextLoader in LangChain

## ما هو `TextLoader`؟
- أبسط وأسرع Document Loader لقراءة الملفات النصية الخام مثل `.txt` و `.md`.
- يقوم بتحميل الملف بالكامل في كائن `Document` واحد يحتوي على:
  - `page_content`: النص الكامل للملف.
  - `metadata`: مسار الملف ومصدره.

</div>


```python
from pathlib import Path
from langchain_community.document_loaders import TextLoader

# تحديد مسار مجلد البيانات بمرونة
DATA_DIR = Path("data") if Path("data").exists() else Path("../../data") if Path("../../data").exists() else Path("../data")

# تحميل وقراءة الملف النصي
file_path = DATA_DIR / "sample.txt"
loader = TextLoader(str(file_path), encoding="utf-8")
docs = loader.load()

print(f"✅ عدد المستندات المحملة: {len(docs)}")
print("--- محتوى المستند ---")
print(docs[0].page_content)
print("\n--- البيانات الوصفية (Metadata) ---")
print(docs[0].metadata)
```
