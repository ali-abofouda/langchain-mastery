# 🎓 04 - ArxivLoader (Academic Papers) in LangChain

## ما هو `ArxivLoader`؟
- أداة بحث واسترجاع للأوراق البحثية العلمية والأكاديمية المنشورة على منصة **arXiv**.
- تقوم بالبحث عن موضوع معين، تنزيل الأوراق ومستخلصاتها وتحويلها إلى مستندات جاهزة للـ RAG.
- تحتوي الـ Metadata على: `Title`, `Authors`, `Published`, `Summary`.

</div>


```python
from langchain_community.document_loaders import ArxivLoader

# البحث عن أحدث ورقة بحثية في موضوع محدد
loader = ArxivLoader(
    query="Retrieval Augmented Generation",
    load_max_docs=1
)

arxiv_docs = loader.load()

print("--- 📚 عنوان البحث ---")
print(arxiv_docs[0].metadata.get("Title"))
print("\n--- ✍️ المؤلفون ---")
print(arxiv_docs[0].metadata.get("Authors"))
print("\n--- 📝 جزء من محتوى البحث ---")
print(arxiv_docs[0].page_content[:400])
```
