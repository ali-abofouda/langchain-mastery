# 🚀 02 - Deploying LCEL Chains with FastAPI & LangServe

## ما هو هذا الكراس؟
يقدم هذا الكراس طريقة تحويل أي **سلسلة LCEL** إلى خادم **REST API** جاهز للاستخدام والتكامل مع واجهات الويب والتطبيقات الذكية باستخدام **FastAPI** و **LangServe**.

</div>

### 1️⃣ تحميل البيئة وتحديد مكونات الخادم

```python
import os
from pathlib import Path
from dotenv import load_dotenv, find_dotenv
from fastapi import FastAPI
from langserve import add_routes
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv(find_dotenv())
print("✅ تم استيراد مكتبات FastAPI و LangServe بنجاح.")
```

### 2️⃣ بناء السلسلة المراد تصديرها كـ API

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "أنت مساعد خبير في البرمجة وتطوير البرمجيات."),
    ("user", "السؤال: {input}")
])

model = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.5
)

chain = prompt | model | StrOutputParser()
print("✅ تم تجهيز السلسلة المعالجة للنشر.")
```

### 3️⃣ إنشاء تطبيق FastAPI وإضافة مسارات LangServe

```python
app = FastAPI(
    title="LangChain API Server",
    version="1.0",
    description="تطبيق خادم خفيف يوفر نقاط اتصال REST API لسلاسل LangChain عبر LangServe"
)

# إضافة المسار للتطبيق
add_routes(
    app,
    chain,
    path="/chain"
)

print("✅ تم إعداد تطبيق FastAPI وإضافة المسار '/chain' بنجاح.")
```
