# 📑 01 - Structured Outputs in LCEL with Pydantic

## ما هو هذا الكراس؟
يشرح هذا الكراس كيفية استخدام **Pydantic** للحصول على مخرجات عالية الدقة وهيكلية (Structured Data) من نماذج المحادثة عبر خاصية `with_structured_output` المدمجة في LangChain.

</div>

### 1️⃣ تحميل البيئة والمستلزمات

```python
import os
from pathlib import Path
from dotenv import load_dotenv, find_dotenv
from pydantic import BaseModel, Field
from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate

load_dotenv(find_dotenv())
print("✅ تم تحميل متغيرات البيئة واستيراد المكتبات بنجاح.")
```

### 2️⃣ تعريف سكيما المخرجات (Pydantic Schema)

```python
class AnswerSchema(BaseModel):
    answer: str = Field(description="الإجابة المباشرة والواضحة على السؤال المطلوب")
    explanation: str = Field(description="شرح موجز يدعم الإجابة المباشرة")
    confidence_score: float = Field(description="مستوى الثقة في الإجابة من 0.0 إلى 1.0")

print("✅ تم تعريف سكيما AnswerSchema باستخدام Pydantic.")
```

### 3️⃣ ربط النموذج وبناء السلسلة الهيكلية

```python
# تهيئة النموذج السريع
llm = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0
)

# تحويل النموذج لإرجاع سكيما محددة
structured_llm = llm.with_structured_output(AnswerSchema)

# إعداد التوجيه
prompt = ChatPromptTemplate.from_template("""
أجب على السؤال التالي واستخرج البيانات وفق السكيما المحددة:

السؤال:
{question}
""")

# بناء السلسلة
chain = prompt | structured_llm

print("✅ تم بناء السلسلة الهيكلية بنجاح.")
```

### 4️⃣ اختبار واستخراج النتائج الهيكلية

```python
question = "ما هي عاصمة فرنسا وما أهميتها؟"
print(f"❓ السؤال: {question}\n")

try:
    result: AnswerSchema = chain.invoke({"question": question})
    print("📊 المخرجات الهيكلية المستخرجة:")
    print(f"  - الإجابة: {result.answer}")
    print(f"  - الشرح: {result.explanation}")
    print(f"  - درجة الثقة: {result.confidence_score}")
except Exception as e:
    print(f"⚠️ خطأ أثناء التشغيل: {e}")
```
