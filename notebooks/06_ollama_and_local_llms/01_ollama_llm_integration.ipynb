# 🦙 01 - Ollama Local LLMs & LangSmith Tracing

## ما هو هذا الكراس؟
يقدم هذا الكراس تطبيقاً عملياً لربط **النماذج المحلية (Local LLMs)** المدارة عبر **Ollama** مع إطار عمل **LangChain**، بالإضافة إلى تفعيل التتبع والمراقبة اللحظية باستخدام **LangSmith**.

## الفوائد والمميزات الرئيسية:
- **التشغيل المحلي الكامل**: الاستغناء عن المفاتيح السحابية وتوفير تكاليف استهلاك الـ Tokens.
- **مرونة التتبع**: متابعة وقت الاستجابة وخطوات تنفيذ السلسلة بالتفصيل على منصة LangSmith.

### 1️⃣ تحميل البيئة وإعداد المتغيرات

```python
import os
from pathlib import Path
from dotenv import load_dotenv, find_dotenv

# تحميل متغيرات البيئة من ملف .env
load_dotenv(find_dotenv())

# ضبط إعدادات التتبع عبر LangSmith
os.environ["LANGSMITH_TRACING"] = "true"
if os.getenv("LANGSMITH_API_KEY"):
    os.environ["LANGSMITH_API_KEY"] = os.getenv("LANGSMITH_API_KEY")
if os.getenv("LANGSMITH_PROJECT"):
    os.environ["LANGSMITH_PROJECT"] = os.getenv("LANGSMITH_PROJECT", "ollama-experiments")

print("✅ تم تحميل متغيرات البيئة وتفعيل تتبع LangSmith بنجاح.")
```

### 2️⃣ تهيئة نموذج Ollama وقالب التوجيه (Prompt Template)

```python
from langchain_ollama import OllamaLLM
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

# إنشاء قالب التوجيه المنظم
prompt = ChatPromptTemplate.from_messages([
    ("system", "أنت مساعد ذكي ومفيد. أجب على الأسئلة بدقة وبصورة مباشرة، وإذا لم تكن تعرف الإجابة قل 'لا أعلم'."),
    ("user", "السؤال: {input}")
])

# تهيئة النموذج المحلي عبر Ollama (مثال: qwen3:8b أو llama3)
llm = OllamaLLM(model="qwen3:8b")
output_parser = StrOutputParser()

print("✅ تم تهيئة نموذج Ollama وقالب التوجيه بنجاح.")
```

### 3️⃣ بناء وتنفيذ سلسلة LCEL

```python
# بناء السلسلة باستخدام عامل الربط |
chain = prompt | llm | output_parser

# استدعاء السلسلة
query = "ما هي أهمية التعلم العميق في الذكاء الاصطناعي؟"
print(f"❓ السؤال: {query}\n")

try:
    response = chain.invoke({"input": query})
    print(f"🤖 الإجابة:\n{response}")
except Exception as e:
    print(f"⚠️ تعذر الاتصال بخادم Ollama المحلي: {e}")
    print("تأكد من تشغيل Ollama محلياً عبر الأمر: ollama run qwen3:8b")
```
