# 🚀 02 - End-to-End Local Ollama Pipeline

## ما هو هذا الكراس؟
هذا هو الكراس الختامي لمسار **النماذج المحلية (Local LLMs)**. يقدم خط أنابيب متكامل يبدأ من تهيئة البيئة والنموذج المحلي عبر **Ollama** وحتى معالجة الاستعلامات وتتبع النتائج بشكل كامل بدون الحاجة لأي خدمات سحابية مدفوعة.

</div>

### 1️⃣ تهيئة البيئة والمكتبات

```python
import os
from pathlib import Path
from dotenv import load_dotenv, find_dotenv
from langchain_ollama import OllamaLLM
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

load_dotenv(find_dotenv())
DATA_DIR = Path("data") if Path("data").exists() else Path("../../data") if Path("../../data").exists() else Path("../data")

print("✅ تم تحميل البيئة وتحديد مجلد البيانات:", DATA_DIR.resolve())
```

### 2️⃣ بناء خط أنابيب الاستجابة التفاعلية (Pipeline Builder)

```python
def create_local_pipeline(model_name: str = "qwen3:8b"):
    prompt = ChatPromptTemplate.from_messages([
        ("system", "أنت خبير محلي في الذكاء الاصطناعي وعلوم البيانات. قدم إجابات ملخصة ودقيقة باللغة العربية."),
        ("user", "الاستفسار: {input}")
    ])
    
    llm = OllamaLLM(model=model_name)
    parser = StrOutputParser()
    
    return prompt | llm | parser

print("✅ تم بناء دالة إنشاء خط الأنابيب المحلي.")
```

### 3️⃣ تشغيل واختبار خط الأنابيب الشامل

```python
pipeline = create_local_pipeline()
test_prompt = "ما هي الخطوات الأساسية لبناء نموذج RAG محلي؟"

print(f"🚀 جاري تشغيل خط الأنابيب عبر النموذج المحلي...\n")
try:
    result = pipeline.invoke({"input": test_prompt})
    print(f"📌 النتيجة الإجمالية:\n{result}")
except Exception as e:
    print(f"⚠️ تعذر التنفيذ المحاكي: {e}")
```
